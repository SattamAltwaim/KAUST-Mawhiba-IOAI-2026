A wine distributor wants two things from their quality-control data:

1. **Predict the exact quality score** (a number from 3 to 9) so they can fine-tune pricing -- this is a **regression** problem.
2. **Flag premium wines** (quality ≥ 7) for a special catalog -- this is a **classification** problem.

The distributor has provided physicochemical measurements (acidity, sugar, alcohol, etc.) for 1,599 red wines. At the bottom of this notebook you will find a model evaluation block that trains both a regressor and a classifier on your prepared data. Your job is to make that block run successfully, achieving:
- **Regression:** R² > 0.28
- **Classification:** Accuracy > 0.78

How you get there is entirely up to you. There are no step-by-step instructions.

**Dataset:** Red Wine Quality (1,599 records, 11 physicochemical features, quality score 3-9)

**Known issues with this dataset:**
- Some features have very different scales (e.g., free sulfur dioxide ranges 1-72, while citric acid ranges 0-1).
- The quality score distribution is heavily concentrated around 5 and 6 -- extreme scores are rare.
- Creating the binary 'premium' target from the quality column requires a threshold choice.
- Some features are moderately correlated with each other.

```python
print("=" * 60)
print("REGRESSION: Ridge")
print("=" * 60)
r2 = r2_score(y_reg_test, ridge_pred)
rmse = np.sqrt(mean_squared_error(y_reg_test, ridge_pred))
mae = mean_absolute_error(y_reg_test, ridge_pred)
print(f"R\u00b2 Score:  {r2:.4f}")
print(f"RMSE:      {rmse:.4f}")
print(f"MAE:       {mae:.4f}")

print()
print("=" * 60)
print("CLASSIFICATION: Logistic Regression")
print("=" * 60)
acc = accuracy_score(y_cls_test, lr_pred)
f1 = f1_score(y_cls_test, lr_pred)
print(f"Accuracy:  {acc:.4f}")
print(f"F1 Score:  {f1:.4f}")
print()
print(classification_report(y_cls_test, lr_pred, target_names=['Standard', 'Premium']))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Regression: predicted vs actual
axes[0].scatter(y_reg_test, ridge_pred, alpha=0.5, edgecolors='k', linewidths=0.3)
lims = [min(y_reg_test.min(), ridge_pred.min()) - 0.5,
        max(y_reg_test.max(), ridge_pred.max()) + 0.5]
axes[0].plot(lims, lims, 'r--', lw=2)
axes[0].set_xlabel('Actual Quality')
axes[0].set_ylabel('Predicted Quality')
axes[0].set_title('Ridge Regression: Predicted vs Actual')

# Classification: confusion matrix
cm = confusion_matrix(y_cls_test, lr_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=axes[1],
            xticklabels=['Standard', 'Premium'],
            yticklabels=['Standard', 'Premium'])
axes[1].set_xlabel('Predicted')
axes[1].set_ylabel('Actual')
axes[1].set_title('Logistic Regression: Confusion Matrix')

plt.tight_layout()
plt.show()
```
